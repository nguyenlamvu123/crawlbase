```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By

import pandas as pd
import time
```

```python
data = pd.read_csv('/home/zaibachkhoa/Downloads/export-as-csv-20241106062517.csv')
data
```

```python
duplicate_rows = data.duplicated(subset=['Tên bài hát'])
data.loc[data.duplicated(subset=['Tên bài hát'])]
data.loc[data['Tên bài hát'] == 'Ngày em đi']
data_ = data.drop_duplicates(subset=['Tên bài hát'])
data_.loc[:, 'Tên bài hát']
data_
```

```python
chrome_service = Service(executable_path='/home/zaibachkhoa/Documents/TikTok/chromedriver-linux64/chromedriver')
driver = webdriver.Chrome(service=chrome_service)
```

```python
# for t2 in tenbaihat_tacgia:
def loop(row):
    driver.get('https://www.tiktok.com/foryou')
    time.sleep(2)
    khungtimkiem = driver.find_element(By.XPATH, '//form[@data-e2e="search-box"]//input')
    khungtimkiem.send_keys(row['Tên bài hát'] + ' ' + row['Tác giả'])
    input('####################################')
    khungtimkiem.send_keys(Keys.RETURN)
    for _ in range(3):
        time.sleep(1)
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    samgiongzon = driver.find_elements(By.XPATH, '//div[@data-e2e="search_top-item"]//a')
    print(len(samgiongzon))
    linklist = list()
    for elem in samgiongzon:
        linklist.append(elem.get_attribute('href'))
        print(linklist[-1])

    for link in linklist:
        time.sleep(1)
        driver.execute_script("window.open('');")
        driver.switch_to.window(driver.window_handles[-1])
        driver.get(link)
        input('press Enter!')
    
        newtab = driver.find_element(By.XPATH, '//h1[@data-e2e="browse-video-desc"]')
        input(data_[data_['Tên bài hát'] == row['Tên bài hát']])
        data_[data_['Tên bài hát'] == row['Tên bài hát']]['2'] = newtab.text
        print('2:', newtab.text)
        newtab = driver.find_element(By.XPATH, '//h4[@data-e2e="browse-music"]')
        data_[data_['Tên bài hát'] == row['Tên bài hát']]['3'] = newtab.text
        print('3:', newtab.text)
    
        time.sleep(2)
        driver.close()
        driver.switch_to.window(driver.window_handles[0])
```
